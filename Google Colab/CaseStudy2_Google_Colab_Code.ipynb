```python
def missing_summary(df):
    summary = pd.DataFrame({
        "Missing Count": df.isnull().sum(),
        "Missing Percentage": (df.isnull().mean() * 100).round(2)
    })

    return summary.sort_values(
        by="Missing Percentage",
        ascending=False
    )


print("========== CUSTOMER DATA ==========")
display(missing_summary(customers))

print("\n========== SALES DATA ==========")
display(missing_summary(sales))

print("\n========== PRODUCT DATA ==========")
display(missing_summary(products))
```

```python
print("========== EXACT DUPLICATE CUSTOMER ID CHECK ==========")

# Keep all rows belonging to duplicated CustomerIDs
duplicate_customers = customers[
    customers["CustomerID"].duplicated(keep=False)
].copy()

# Check whether each duplicated CustomerID has only one unique
# combination of the remaining customer attributes
attribute_columns = [
    "FirstName",
    "LastName",
    "Gender",
    "Region",
    "SSN"
]

duplicate_consistency = (
    duplicate_customers
    .groupby("CustomerID")[attribute_columns]
    .nunique(dropna=False)
)

# A value of 1 in every attribute means the duplicate records
# contain the same customer information
exact_duplicates = duplicate_consistency.eq(1).all(axis=1)

print("Duplicated CustomerIDs:", len(duplicate_consistency))
print("Exact duplicate CustomerIDs:", exact_duplicates.sum())
print("Conflicting CustomerIDs:", (~exact_duplicates).sum())

print("\n========== CONFLICTING CUSTOMER IDs ==========")
display(duplicate_consistency[~exact_duplicates])
```

```python
print("========== SALES AMOUNT OUTLIER CHECK ==========")

sales_amount = sales["SalesAmount"].dropna()

Q1 = sales_amount.quantile(0.25)
Q3 = sales_amount.quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = sales[
    (sales["SalesAmount"] < lower_bound) |
    (sales["SalesAmount"] > upper_bound)
]

print("Q1:", round(Q1, 2))
print("Q3:", round(Q3, 2))
print("IQR:", round(IQR, 2))

print("\nLower Bound:", round(lower_bound, 2))
print("Upper Bound:", round(upper_bound, 2))

print("\nNumber of potential outliers:", len(outliers))

print("\n========== TOP SALES AMOUNT VALUES ==========")
display(
    sales[["SaleID", "ProductID", "CustomerID", "SalesAmount", "Quantity"]]
    .sort_values("SalesAmount", ascending=False)
    .head(10)
)
```

```python
print("========== SALES AMOUNT OUTLIER ANALYSIS ==========")

# Add ProductName and Category to the outlier records
outlier_analysis = outliers.merge(
    products[["ProductID", "ProductName", "Category"]],
    on="ProductID",
    how="left"
)

print("Potential outlier transactions:", len(outlier_analysis))

print("\n========== OUTLIERS BY CATEGORY ==========")

category_outliers = (
    outlier_analysis["Category"]
    .value_counts()
    .rename_axis("Category")
    .reset_index(name="Outlier Count")
)

display(category_outliers)

print("\n========== OUTLIER SALES BY PRODUCT ==========")

product_outliers = (
    outlier_analysis
    .groupby(
        ["ProductID", "ProductName", "Category"],
        as_index=False
    )
    .agg(
        OutlierTransactions=("SaleID", "count"),
        TotalOutlierSales=("SalesAmount", "sum"),
        MaxSalesAmount=("SalesAmount", "max")
    )
    .sort_values("TotalOutlierSales", ascending=False)
)

display(product_outliers.head(15))
```

```python
print("========== MISSING SALES DATA BY PRODUCT ==========")

missing_sales = sales[
    sales["SalesAmount"].isna() | sales["Quantity"].isna()
].copy()

product_missing_profile = (
    missing_sales
    .groupby("ProductID")
    .agg(
        MissingSalesAmount=("SalesAmount", lambda x: x.isna().sum()),
        MissingQuantity=("Quantity", lambda x: x.isna().sum()),
        TotalMissingRows=("ProductID", "size")
    )
    .reset_index()
    .sort_values("TotalMissingRows", ascending=False)
)

print(
    "Products affected by missing SalesAmount or Quantity:",
    len(product_missing_profile)
)

display(product_missing_profile.head(20))

print("\n========== MISSINGNESS PERCENTAGE BY PRODUCT ==========")

product_total_counts = (
    sales["ProductID"]
    .value_counts()
    .rename("TotalSalesRows")
    .reset_index()
    .rename(columns={"index": "ProductID"})
)

product_missing_profile = product_missing_profile.merge(
    product_total_counts,
    on="ProductID",
    how="left"
)

product_missing_profile["MissingRowPercentage"] = (
    product_missing_profile["TotalMissingRows"]
    / product_missing_profile["TotalSalesRows"]
    * 100
)

display(
    product_missing_profile
    .sort_values("MissingRowPercentage", ascending=False)
    .head(20)
)
```

```python
print("========== PHASE 2: STEP 23 — MYSQL CREATE TABLE DDL ==========")

mysql_ddl = """

CREATE DATABASE IF NOT EXISTS retail_sales_dw;

USE retail_sales_dw;


-- ============================================================
-- DIM CUSTOMER
-- ============================================================

CREATE TABLE IF NOT EXISTS DimCustomer (
    CustomerID VARCHAR(4) PRIMARY KEY,
    FirstName VARCHAR(100) NOT NULL,
    LastName VARCHAR(100) NULL,
    Gender VARCHAR(20) NOT NULL,
    Region VARCHAR(50) NOT NULL,
    SSN CHAR(9) NOT NULL
);


-- ============================================================
-- DIM PRODUCT
-- ============================================================

CREATE TABLE IF NOT EXISTS DimProduct (
    ProductID INT PRIMARY KEY,
    ProductName VARCHAR(100) NOT NULL,
    Category VARCHAR(50) NOT NULL
);


-- ============================================================
-- DIM DATE
-- ============================================================

CREATE TABLE IF NOT EXISTS DimDate (
    DateKey INT PRIMARY KEY,
    FullDate DATE NOT NULL,
    Year INT NOT NULL,
    MonthNumber INT NOT NULL,
    MonthName VARCHAR(20) NOT NULL,
    Quarter VARCHAR(2) NOT NULL,
    Day INT NOT NULL,
    DayName VARCHAR(20) NOT NULL,
    DayOfWeek INT NOT NULL
);


-- ============================================================
-- FACT SALES
-- ============================================================

CREATE TABLE IF NOT EXISTS FactSales (
    SaleID INT PRIMARY KEY,
    SourceSaleID VARCHAR(36) NOT NULL UNIQUE,
    ProductID INT NOT NULL,
    CustomerID VARCHAR(4) NOT NULL,
    DateKey INT NOT NULL,
    SalesAmount DECIMAL(10,2) NOT NULL,
    Quantity INT NOT NULL,
    Timestamp DATETIME NOT NULL,

    CONSTRAINT fk_factsales_customer
        FOREIGN KEY (CustomerID)
        REFERENCES DimCustomer(CustomerID),

    CONSTRAINT fk_factsales_product
        FOREIGN KEY (ProductID)
        REFERENCES DimProduct(ProductID),

    CONSTRAINT fk_factsales_date
        FOREIGN KEY (DateKey)
        REFERENCES DimDate(DateKey)
);

"""

print(mysql_ddl)


# ============================================================
# BASIC DDL DESIGN VALIDATION
# ============================================================

required_tables = [
    "DimCustomer",
    "DimProduct",
    "DimDate",
    "FactSales"
]

required_foreign_keys = [
    "fk_factsales_customer",
    "fk_factsales_product",
    "fk_factsales_date"
]

print("\n========== DDL DESIGN CHECK ==========")

for table in required_tables:
    print(f"{table}: INCLUDED")

print("\nForeign Keys:")

for fk in required_foreign_keys:
    print(f"{fk}: INCLUDED")


# ============================================================
# IMPORTANT COLUMN CHECKS
# ============================================================

ddl_checks = [
    "CustomerID VARCHAR(4)",
    "SourceSaleID VARCHAR(36)",
    "SalesAmount DECIMAL(10,2)",
    "Quantity INT",
    "DateKey INT",
    "FOREIGN KEY (CustomerID)",
    "FOREIGN KEY (ProductID)",
    "FOREIGN KEY (DateKey)"
]

print("\n========== KEY DDL DEFINITIONS ==========")

for check in ddl_checks:
    print(
        check,
        ":",
        "FOUND" if check in mysql_ddl else "MISSING"
    )


ddl_status = all(
    check in mysql_ddl
    for check in ddl_checks
)

print("\n========== MYSQL DDL STATUS ==========")

print(
    "CREATE TABLE DDL validation:",
    "PASS" if ddl_status else "FAIL"
)
```

```python
print("========== PHASE 2: STEP 25 — CREATE MYSQL IMPORT PACKAGE ==========")

import os
import zipfile

output_dir = "/content/retail_sales_mysql"

zip_path = "/content/retail_sales_mysql_import.zip"

# ============================================================
# CREATE ZIP PACKAGE
# ============================================================

with zipfile.ZipFile(
    zip_path,
    mode="w",
    compression=zipfile.ZIP_DEFLATED
) as zip_file:

    for filename in sorted(os.listdir(output_dir)):

        filepath = os.path.join(output_dir, filename)

        zip_file.write(
            filepath,
            arcname=filename
        )


# ============================================================
# VERIFY ZIP CONTENTS
# ============================================================

print("\n========== ZIP CONTENTS ==========")

with zipfile.ZipFile(zip_path, "r") as zip_file:

    files_inside = zip_file.namelist()

    for filename in files_inside:
        print(filename)


# ============================================================
# FILE COUNT
# ============================================================

print("\n========== PACKAGE VALIDATION ==========")

expected_files = {
    "DimCustomer.csv",
    "DimProduct.csv",
    "DimDate.csv",
    "FactSales.csv",
    "retail_sales_dw.sql"
}

actual_files = set(files_inside)

missing_files = expected_files - actual_files
unexpected_files = actual_files - expected_files

print("Expected files:", len(expected_files))
print("Files in package:", len(actual_files))
print("Missing files:", len(missing_files))
print("Unexpected files:", len(unexpected_files))

if missing_files:
    print("Missing:", sorted(missing_files))

if unexpected_files:
    print("Unexpected:", sorted(unexpected_files))


# ============================================================
# ZIP SIZE
# ============================================================

zip_size_kb = os.path.getsize(zip_path) / 1024

print("\nZIP package size:",
      f"{zip_size_kb:.2f} KB")


# ============================================================
# FINAL STATUS
# ============================================================

package_status = (
    actual_files == expected_files
)

print("\n========== MYSQL IMPORT PACKAGE STATUS ==========")

print(
    "MySQL import package:",
    "PASS" if package_status else "FAIL"
)
```
